## List Comprehension



new_list = [expression for item in iterable if condition]


expression → The operation or value to include in the new list.


item → The variable representing each element from the iterable.


iterable → The source of elements (e.g., list, range, string, etc.).


if condition (optional) → Filters elements before adding them.


```python
l3 = [num for num in range(1,11)]
l3
```

```python
l4 = [char for char in 'Abishek']
l4
```

```python
# taking only consonents from a string

l5 = [ch for ch in 'Abishek' if ch not in 'aeiouAEIOU']
l5
```

```python
# Even numbers

l6 =[num for num in range(0,50) if num%2 ==0]

l6
```

when we have if and else block the syntax is

new_list = [expression_if_true if condition else expression_if_false for item in iterable]

```python
l7 = [num if num%2==0 else num+100 for num in range(1,10)]
l7
```

```python
l8 = [ch if ch in 'aeiouAEIOU' else f'{ch} is not a vowel' for ch in 'Abishek']
l8
```

## Nested list Comprehension

```python
[[i*2 for i in range(num)] for num in range(1,5)]
```

```python
[ num for num in range(5)]
```

In place of the returning experssion we will write another list comprehension to make it nested list comprehension

```python
[   [i]     for i in range(5)]
```

```python
[   [num2 for num2 in range(10,13)]     for num in range(5)]
```
