## Break Statement

```python
start = -2
end = 2
num = 10
while (start <= end):
    print(f'Answer for start = {start}')
    if start == 0:
        print('Loop Completed')
        break
    print(num/start)
    start = start +1
    
```

## Continue Staement

```python
# printing odd number
for num in range(1,11):
    if num % 2 == 0:
        continue
    print(num)
    
```
