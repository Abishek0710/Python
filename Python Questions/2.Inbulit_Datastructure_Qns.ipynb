```python
"""
Sum of List Elements

Example:
        Input: numbers = [1, 2, 3, 4, 5]
        Output: 15

        Input: numbers = [10, -5, 7, 8, -2]
        Output: 18

"""


def sum_list(numbers):
    # Your code goes here
    ans_sum = 0
    for i in numbers:
        ans_sum= ans_sum + i
        
    return ans_sum


numbers = [1, 2, 3, 4, 5]
sum_list(numbers)
```

```python
""""

Largest Element in a List

Example:

        Input: numbers = [3, 8, 2, 10, 5]
        Output: 10

        Input: numbers = [-5, -10, -2, -1, -7]
        Output: -1

"""


def find_largest(numbers):
    
    largest = numbers[0]
    
    for num in numbers[1:]:
        
        if num > largest:
            largest = num
            
    return largest

find_largest([3, 8, 2, 10, 5])
```

```python
"""
Remove Duplicate in a List

Example:

        Input: lst = [1, 2, 2, 3, 4, 4, 5]
        Output: [1, 2, 3, 4, 5]

        Input: lst = [4, 5, 5, 4, 6, 7]
        Output: [4, 5, 6, 7]
"""

def remove_duplicates(lst):
    
    ans = []
    
    for num in lst:
        
        if num not in ans:
            ans.append(num)
            
    return ans
    
    
remove_duplicates([1, 2, 2, 3, 4, 4, 5])
```

```python
"""

Check if all elements in a list are Unique

Example:

        Input: lst = [1, 2, 3, 4, 5]
        Output: True

        Input: lst = [1, 2, 3, 3, 4, 5]
        Output: False
"""

def check_unique(lst):
    
    
    ans =[]
    
    for num in lst:
        
        if num in ans:
            
            return False
        else:
            ans.append(num)
    
    return True

check_unique([1, 2, 3, 4, 5])
```

```python
"""
Program to Reverse a List

Example:

Input: lst = [1, 2, 3, 4, 5]
Output: [5, 4, 3, 2, 1]
"""

def reverse_list(lst):
    
    ans = []
    
    for i in lst[::-1]:
        
        ans.append(i)
    return ans
        
    
reverse_list([1, 2, 3, 4, 5])
```

```python
"""
Count Number of Odd and Even Elements in a List

Example:
    Input: lst = [1, 2, 3, 4, 5]
    Output: (2, 3)

    There are 2 even numbers: 2, 4

    There are 3 odd numbers: 1, 3, 5
"""

def count_even_odd(lst):
    
    
    odd = 0
    even = 0
    
    for num in lst:
        
        if num%2==0:
            even = even + 1
        else:
            odd = odd + 1 
    return (even, odd)

count_even_odd([1, 2, 3, 4, 5])
```

```python
"""
Maximum difference between two consecutive elements in a list.

Example:
        Input: lst = [1, 7, 3, 10, 5]
        Output: 7

        The maximum difference is between 3 and 10 (i.e., |3 - 10| = 7).

        Input: lst = [10, 11, 15, 3]
        Output: 12

        The maximum difference is between 15 and 3 (i.e., |15 - 3| = 12).
"""

def max_consecutive_difference(lst):
    
    if len(lst)< 2:
        return 0
    
    
    diff = 0 
    for i in range(len(lst)-1):
        temp = abs(lst[i] - lst[i+1])
        if temp > diff:
            diff = temp
        
    
    return diff

max_consecutive_difference([1, 7, 3, 10, 5])
```
