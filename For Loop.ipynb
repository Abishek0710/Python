## Range Function

```python
# range starts from 0 and goes in forward direction it cannot go in negative direction
r1 = range(5)
```

```python
r1
```

```python
list(r1)
```

```python
# range of negative value is empty
list(range(-4))
```

```python
# even if its negative it tends to go in positive direction
(range(-10,-5))
```

```python
list(range(-5,-10))
```

```python
# range with 3 parameters
# 3rd parameter is step
list(range(0,10,2))
```

```python
# range does not move in negative direction but we can make it move using step parameter

list(range(-5,-10,-1))
```

```python
# float value in step is not supported also step cannot be 0
list(range(0,5,0.5))
```

```python
list(range(0,5,0))
```

```python
# sum off numbers

number = int(input('Enter a whole number: '))

sum = 0

for i in range(0,number):
    sum = sum + i
    print(sum)
print('----------')
print(sum)
```

```python
# Factorial of a number
number = int(input("Enter a number: "))

fact = 1

for i in range(1, number+1):
    fact =fact *i
    print(fact)
    
print('---------')

print(fact)
    
```

## Nested Loop

```python
for i in range(1,11):
    print('Value of i outside loop:   ',i)
    for j in range(3,5):
        print( 'Value of j in inner loop:  ', j)
        print(f'multiplication of i*j is :   {i} *{j}:   ', i*j)
    print('value of i outside inner loop:    ', i)
    print('----------------')
```

```python
for i in range(1,11):
    print('Value of i outside loop:   ',i)
    if i==5:
        break
    for j in range(3,5):
        print( 'Value of j in inner loop:  ', j)
        print(f'multiplication of i*j is :   {i} *{j} :   ', i*j)
    print('value of i outside inner loop:    ', i)
    print('----------------')
```

```python
n = 5
for outer in range(1,n+1):
    print(outer)
```

```python
for outer in range(1,n+1):
    for inner in range(outer):
        print(outer)
```

```python
for outer in range(1,n+1):
    for inner in range(outer):
        print(outer,end='')
```
